## 6. Cuartiles: Calcula los cuartiles de tip.

```python
y = df_tips.tip
y
```

```python
np.quantile(y,[0.25,0.5,0.75])
```

```python
y.quantile([0.25,0.5,0.75])
```

```python
df_tips.describe()
```

## 7. Diagrama de cajas: Visualiza tip y comenta tus observaciones.

```python
q1,q2,q3=y.quantile([0.25, 0.50, 0.75])
```

```python
q1, q2, q3 # serie con los valores de q1, q2, q3
```

```python
iqr = q3 - q1  #Calculé el IQR usando pandas (q1 -> cuantil 25% -primer cuartil-, q3 -> cuantil 75% -tercer cuartil-) de la columna 'Tip'
print(f'Rango Intercuartílico IQR = {iqr} ')
```

```python
plt.figure(figsize=(6, 2))
plt.boxplot(y, vert=False, patch_artist=True, boxprops=dict(facecolor='lightgreen'))
plt.title('Diagrama de Cajas (Boxplot)')
plt.xlabel('Tip')
plt.show()
```
