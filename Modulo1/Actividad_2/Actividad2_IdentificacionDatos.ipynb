## Datos Atípicos (Outliers) Multivariado
1) Local Outlier Factor (LOF)

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
#Modifique el archivo creditcard.csv, sólo deje 214 registros, demasiado pesado para procesar por mi equipo.
```

```python
df_cc = pd.read_csv('creditcard.csv')
df_cc
```

```python
df_cc.head(5)
```

```python
clf = LocalOutlierFactor(n_neighbors=20, contamination='auto')
X = df_cc[['V1','V2']].values
y_pred = clf.fit_predict(X)
```

```python
y_pred
```

```python
plt.figure(figsize=(9,6))
in_mask = [True if l == 1 else False for l in y_pred]
out_mask = [True if l == -1 else False for l in y_pred]

plt.title("Detección de Outliers en Tarjetas de Credito con Local Outlier Factor (LOF)")

a = plt.scatter(X[in_mask, 0], X[in_mask, 1], c = 'green',
                edgecolor = 'k', s = 30)
# outliers
b = plt.scatter(X[out_mask, 0], X[out_mask, 1], c = 'orange',
                edgecolor = 'k', s = 30)
plt.axis('tight')
plt.xlabel('V1');
plt.ylabel('V2');
plt.show()
```

2) Isolation Forest (IF)

```python
from sklearn.ensemble import IsolationForest
```

```python
#Seleccionamos las mismas variables que en LOF, para detectar outlier
X_real = df_cc[['V1', 'V2']].values
```

```python
#Declaración de los limites - parametros
clf_real = IsolationForest(contamination=0.1)   #10% datos outliers
y_pred_real = clf_real.fit_predict(X_real)
```

```python
#Graficar Resultados
plt.figure(figsize=(9,6))
plt.scatter(X_real[:,0], X_real[:,1], c=(y_pred_real==-1), cmap='Greens', s=14, alpha=0.5)
plt.xlabel('V1')
plt.ylabel('V2')
plt.title('Detección de Outliers en Tarjetas de Credito con Isolation Forest')
plt.show()
```

3) Density-Based Spatial Clustering of Applications with Noise (DBSCAN)

```python
from sklearn.cluster import DBSCAN
```

```python
Y = df_cc[[f'V{i}' for i in range(1,10)]].values

db = DBSCAN(eps=1.8, min_samples=20).fit(Y)
labels = db.labels_
```

```python
pd.Series(labels).value_counts()
```

```python
unique_labels = set(labels)
colors = plt.cm.tab10(np.linspace(0,1,len(unique_labels)))

plt.figure(figsize=(8,6))
for color,label in zip(colors, unique_labels):
    sample_mask = [True if l == label else False for l in labels]
    plt.plot(Y[:,0][sample_mask], Y[:, 1][sample_mask], '*', color=color);
plt.xlabel('Time');
plt.ylabel('Amount');
plt.show()
```
